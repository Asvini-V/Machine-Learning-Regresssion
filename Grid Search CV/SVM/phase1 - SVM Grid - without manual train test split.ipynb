```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
dataset=pd.read_csv("insurance_pre.csv")
```

```python
dataset
```

```python
dataset=pd.get_dummies(dataset, drop_first=True)
```

```python
dataset
```

```python
independent= dataset[['age', 'bmi', 'children', 'sex_male', 'smoker_yes']]
independent
```

```python
dependent= dataset['charges']
dependent
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.model_selection import GridSearchCV
from sklearn.svm import SVR

# Define the Pipeline (Scales data then applies SVR)
pipe = Pipeline([
    ('scaler', StandardScaler()),
    ('svr', SVR())
])
```

```python
# Update param_grid to match Pipeline keys (note the 'svr__' prefix)
param_grid = {
    'svr__kernel': ['rbf', 'poly', 'sigmoid', 'linear'],
    'svr__C': [10, 100, 1000, 2000, 3000],
    'svr__gamma': ['auto', 'scale']
}
```

```python
#  Fit on the entire dataset (independent, dependent)
grid = GridSearchCV(pipe, param_grid, refit=True, verbose=3, n_jobs=1) # n_jobs=-1 makes it faster
grid.fit(independent, dependent)
```

```python
# Results
print("Best Parameters:", grid.best_params_)
print("Best Cross-Validated R² Score:", grid.best_score_)
table = pd.DataFrame.from_dict(grid.cv_results_)
```

```python
# Prediction (The grid object will now automatically scale your new input!)
age_input = float(input("Age:"))
bmi_input = float(input("BMI:"))
children_input = float(input("Children:"))
sex_male_input = int(input("Sex Male 0 or 1:"))
smoker_yes_input = int(input("Smoker Yes 0 or 1:"))
```

```python
# No need to manually scale this; the pipeline handles it
Future_Prediction = grid.predict([[age_input, bmi_input, children_input, sex_male_input, smoker_yes_input]])
print("Future_Prediction={}".format(Future_Prediction))
```
