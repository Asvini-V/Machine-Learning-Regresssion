```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
dataset=pd.read_csv("insurance_pre.csv")
dataset
```

```python
dataset=pd.get_dummies(dataset, drop_first=True)
```

```python
independent= dataset[['age', 'bmi', 'children', 'sex_male', 'smoker_yes']]
independent
```

```python
dependent= dataset['charges']
dependent
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.tree import DecisionTreeRegressor
param_grid = {'criterion':['squared_error','absolute_error','friedman_mse'], 
              'max_features': ['auto','sqrt','log2'],
              'splitter':['best','random']}
```

```python
grid = GridSearchCV(DecisionTreeRegressor(), param_grid, refit = True, verbose= 3,n_jobs=1)
# fitting the model for grid search
grid.fit(independent, dependent)
```

```python
# Best results
print("Best Parameters:", grid.best_params_)
print("Best Cross-Validated R² Score:", grid.best_score_)
```

```python
# Store CV results in table
results_table = pd.DataFrame(grid.cv_results_)
results_table
```

```python
# Future prediction
age_input = float(input("Age: "))
bmi_input = float(input("BMI: "))
children_input = float(input("Children: "))
sex_male_input = int(input("Sex Male (0 or 1): "))
smoker_yes_input = int(input("Smoker Yes (0 or 1): "))
```

```python
future_prediction = grid.best_estimator_.predict(
    [[age_input, bmi_input, children_input, sex_male_input, smoker_yes_input]]
)
print("Predicted Insurance Charges:", future_prediction)
```
