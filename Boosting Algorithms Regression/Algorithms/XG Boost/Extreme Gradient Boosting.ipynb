```python
import pandas as pd
```

```python
dataset=pd.read_csv("insurance_pre.csv")
```

```python
dataset=pd.get_dummies(dataset,drop_first=True)
```

```python
independent= dataset[['age', 'bmi', 'children', 'sex_male', 'smoker_yes']]
independent
```

```python
dependent= dataset[['charges']]
dependent
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(independent, dependent, test_size=0.30, random_state=0)
```

```python
# !pip install xgboost
```

```python
from xgboost import XGBRegressor

regressor = XGBRegressor(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=3,
    random_state=0
)

regressor.fit(X_train, y_train)
```

```python
y_pred=regressor.predict(X_test)
```

```python
from sklearn.metrics import r2_score 
r_score=r2_score(y_test,y_pred)
r_score
```

```python
import pickle 
filename="finalized_model_XG-Boost.sav"
```

```python
pickle.dump(regressor,open(filename,'wb'))
```

```python
loaded_model=pickle.load(open("finalized_model_XG-Boost.sav", 'rb')) 
row_to_predict = pd.DataFrame([[48, 25.856, 3, 1, 0]], 
                             columns=['age', 'bmi', 'children', 'sex_male', 'smoker_yes'])

# 4. Predict
result = loaded_model.predict(row_to_predict)
# result
print(f"Predicted Insurance Charge: {result[0]}")
```
