```python
# Pandas is a Python library used to read and work with data.
# pd is just a short name used to make coding easier.

import pandas as pd
```

```python
# This line reads the salary data from a CSV file and stores it in a variable called dataset.

dataset=pd.read_csv("Salary_Data.csv")
```

```python
# This command displays the dataset.

dataset
```

```python
# Here the input column YearsExperience is selected and stored in a variable named independent. 

independent=dataset[["YearsExperience"]]
independent
```

```python
# Here the output column Salary is selected and stored in a variable named dependent. 

dependent=dataset[["Salary"]]
dependent
```

```python
# sklearn is a machine learning library in Python used to build, train, and test models easily.
# Here, using the sklearn library the dataset is split into training and testing data.
# 70% of the data is used to train the model and 30% is used to test it.

from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(independent, dependent, test_size=0.30, random_state=0)
```

```python
# A Linear Regression model is created and trained using the training data.

from sklearn.linear_model import LinearRegression # imports the Linear Regression model from the sklearn library.
regressor=LinearRegression() #  creates a Linear Regression model and stores it in the variable named regressor.
regressor.fit(X_train,y_train) # trains the model using the training input (X_train) and output (y_train) data.
```

```python
weight=regressor.coef_  #  This line gets the weight (coefficient) value from the trained model.
weight # displays the weight value
```

```python
bias=regressor.intercept_  # This line gets the bias (intercept) value from the trained model.
bias # displays the bias value
```

```python
# Predicts the salary for the test data using the trained model.

y_pred=regressor.predict(X_test)
```

```python
from sklearn.metrics import r2_score # Imports the r2_score function from the sklearn.metrics module
r_score=r2_score(y_test,y_pred) # Computes the R² score by comparing actual and predicted values.
```

```python
r_score # Displays the R² score of the model.
```

```python
import pickle # Imports the pickle module, used to save Python objects (like models).
filename="finalized_model_linear.sav" # Defines the filename to save the trained model.
```

```python
pickle.dump(regressor,open(filename,'wb')) # Saves the trained model (regressor) to a file in binary mode ('wb')
                                           # using pickle.dump().
```

```python
loaded_model=pickle.load(open("finalized_model_linear.sav", 'rb')) # Loads the saved model from the file in read-binary mode ('rb') 
                                                                   # using pickle.load()
result=loaded_model.predict([[13]]) # Uses the loaded model to predict the salary for 13 years of experience.
```
